***Algoritmo genético que busca optimizar una población para encontrar el mejor individuo a partir de un modelo definido***

```python
import random as rd
```

**Definir individuo**

```python
def getIndividuo(min,max,tamano):
  return [rd.randint(min,max) for x in range(tamano)]
```

**Definir población**

```python
def getPoblacion(numero):
  return [getIndividuo(0,valorMaximo,tamano) for i in range (numero)]
```

**Calcular adaptación** (Función objetivo)

```python
def CalcularAdaptacion(individuo):
  valor=0
  for i in range(len(individuo)):
    if modelo[i]==individuo[i]:
      valor+=1
  return valor
```

**Realizar Cruce**

```python
def cruce(lPoblacion,lPadres):
  for i in range(len(lPoblacion)-1):
    puntoCruce = rd.randint(0, tamano -1)
    #print("punto cruce %s" % (puntoCruce))
    lPoblacion[i][:puntoCruce]=lPadres[0][1][:puntoCruce]
    lPoblacion[i][puntoCruce:]=lPadres[1][1][puntoCruce:]
  return lPoblacion
```

**Realizar Mutación**

```python
def mutacion(listaCruzados):
  for i in range(len(listaCruzados)-1):
     if rd.random() <= probMutacion:
      punto = rd.randint(0, tamano-1)
      nuevo_valor = rd.randint(0, valorMaximo)
      while nuevo_valor == listaCruzados[i][punto]:
       nuevo_valor = rd.randint(0, valorMaximo)
      listaCruzados[i][punto]=nuevo_valor
  return listaCruzados
```

**Método Ver**

```python
def verLista(lista, texto):
  print("\n"+texto)
  for i in lista:
    print (i)
```

**Flujo del algoritmo genético**

```python
tamano=8
tamanoPoblacion=8
valorMaximo=8
numMejores=5
iteraciones=500
probMutacion=0.8

modelo=getIndividuo(0,valorMaximo,tamanoPoblacion)
print("Modelo %s" % (modelo))

#-----------Generar la población ------------------
listaPoblacion = getPoblacion(tamanoPoblacion)
verLista(listaPoblacion, "---Población inicial ---")

for j in range(iteraciones):
  #---------Calculo de adaptación---------
  listaValorados = [(CalcularAdaptacion(i),i) for i in listaPoblacion]
  listaOrdenados=sorted(listaValorados, reverse=True)
  #verLista(listaOrdenados, "---Población Valorada ---")
  #print(listaOrdenados[0][0])

  #---------Selección---------"
#  listaSeleccionados = listaOrdenados[:(len(listaOrdenados) - numMejores)]
  listaSeleccionados = listaOrdenados[:numMejores]
  listaPadres=rd.sample(listaSeleccionados,2)
  #verLista(listaPadres, "---Individuos seleccionados como padres ---")

  #---------Cruce-------------"
  listaPoblacionCruzada=cruce(listaPoblacion,listaPadres)
  #verLista(listaPoblacionCruzada,"---Población Cruzada ---")

  #---------Mutación-------------"
  listaPoblacionMutada=mutacion(listaPoblacionCruzada)
  #verLista(listaPoblacionMutada,"---Población Mutada ---")
  listaPoblacion=listaPoblacionMutada

#verLista(listaPoblacion,"---Población Final ---")

#---mejor individuo---
listaMejores = [(CalcularAdaptacion(i),i) for i in listaPoblacion]
listaMejores=sorted(listaMejores, reverse=True)
print("\nMejor individuo después de "+ str(iteraciones) + " iteraciones")
print(listaMejores[0])
```
